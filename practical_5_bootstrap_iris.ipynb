# Practical 5: Bootstrap resampling on Iris dataset

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

np.random.seed(42)
iris = load_iris(as_frame=True).frame
data = iris["sepal length (cm)"].values

n_boot = 1000
boot_means = np.array([
    np.random.choice(data, size=len(data), replace=True).mean()
    for _ in range(n_boot)])

ci_low, ci_high = np.percentile(boot_means, [2.5, 97.5])
print("Original mean :", round(data.mean(), 4))
print("Bootstrap mean:", round(boot_means.mean(), 4))
print("Standard error:", round(boot_means.std(), 4))
print(f"95% CI: ({ci_low:.4f}, {ci_high:.4f})")

plt.hist(boot_means, bins=30, edgecolor="black", alpha=0.7)
plt.axvline(ci_low, color="r", ls="--", label="95% CI")
plt.axvline(ci_high, color="r", ls="--")
plt.axvline(data.mean(), color="k", label="Original mean")
plt.title("Bootstrap distribution of mean sepal length")
plt.legend()
plt.show()
```
