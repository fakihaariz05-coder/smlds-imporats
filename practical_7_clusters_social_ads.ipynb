# Practical 7: Plot clusters on Social_Network_Ads.csv (K-Means)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

df = pd.read_csv("Social_Network_Ads.csv")
X = df[["Age", "EstimatedSalary"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# elbow method to choose k
wcss = [KMeans(n_clusters=i, n_init=10, random_state=42).fit(X_scaled).inertia_
        for i in range(1, 11)]
plt.plot(range(1, 11), wcss, "o-")
plt.title("Elbow method")
plt.xlabel("Number of clusters (k)")
plt.ylabel("WCSS")
plt.show()
```

```python
# final clustering
k = 3
km = KMeans(n_clusters=k, n_init=10, random_state=42)
df["Cluster"] = km.fit_predict(X_scaled)
centers = scaler.inverse_transform(km.cluster_centers_)

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="Age", y="EstimatedSalary",
                hue="Cluster", palette="Set1", s=50)
plt.scatter(centers[:, 0], centers[:, 1], c="black", marker="X", s=200, label="Centroids")
plt.title(f"K-Means clusters (k={k})")
plt.legend()
plt.show()
```
