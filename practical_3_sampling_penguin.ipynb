# Practical 3: Simple and stratified sampling on penguin dataset

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins").dropna()
print("Population size:", len(penguins))
```

```python
# simple random sampling (10%)
simple = penguins.sample(frac=0.10, random_state=42)
```

```python
# stratified sampling (10% from each species)
stratified = penguins.groupby("species", group_keys=False).sample(
    frac=0.10, random_state=42)

print("Simple sample size:", len(simple))
print("Stratified sample size:", len(stratified))

comparison = pd.DataFrame({
    "Population": penguins["species"].value_counts(normalize=True),
    "Simple": simple["species"].value_counts(normalize=True),
    "Stratified": stratified["species"].value_counts(normalize=True),
})
print("\nSpecies proportions:\n", comparison.round(3))

comparison.plot(kind="bar", figsize=(8, 5))
plt.title("Species proportion: population vs samples")
plt.ylabel("Proportion")
plt.show()
```
