# Practical 4: Central Limit Theorem on exponential data

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
population = np.random.exponential(scale=2.0, size=100_000)  # skewed data
sample_sizes = [1, 5, 30, 100]

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, n in zip(axes, sample_sizes):
    # 1000 samples of size n -> mean of each
    means = [np.random.choice(population, n).mean() for _ in range(1000)]
    ax.hist(means, bins=30, density=True, edgecolor="black", alpha=0.7)

    # theoretical normal curve predicted by CLT
    mu, sigma = population.mean(), population.std() / np.sqrt(n)
    x = np.linspace(min(means), max(means), 200)
    ax.plot(x, stats.norm.pdf(x, mu, sigma), "r-", lw=2)
    ax.set_title(f"n = {n}")

plt.suptitle("CLT: distribution of sample means (exponential data)")
plt.tight_layout()
plt.show()
```
