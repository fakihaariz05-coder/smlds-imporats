# Practical 6: Maximum Likelihood Estimation on Iris dataset

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True).frame
data = iris["sepal length (cm)"].values
```

```python
# MLE formulas for Normal distribution
mu_mle = data.mean()
sigma2_mle = np.mean((data - mu_mle) ** 2)   # divide by n (not n-1)
sigma_mle = np.sqrt(sigma2_mle)
print(f"MLE mean = {mu_mle:.4f}")
print(f"MLE std  = {sigma_mle:.4f}")
```

```python
# cross-check with scipy
mu_s, sigma_s = stats.norm.fit(data)
print(f"scipy fit: mean = {mu_s:.4f}, std = {sigma_s:.4f}")
```

```python
# log-likelihood at the MLE
loglik = np.sum(stats.norm.logpdf(data, mu_mle, sigma_mle))
print("Log-likelihood:", round(loglik, 4))
```

```python
# plot fitted curve
x = np.linspace(data.min() - 1, data.max() + 1, 200)
plt.hist(data, bins=15, density=True, alpha=0.6, edgecolor="black")
plt.plot(x, stats.norm.pdf(x, mu_mle, sigma_mle), "r-", lw=2, label="MLE Normal fit")
plt.title("MLE fit on Iris sepal length")
plt.legend()
plt.show()
```
